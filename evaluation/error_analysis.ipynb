```python
# evaluation/error_analysis.ipynb

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# scikit-learn confusion matrix
from sklearn.metrics import confusion_matrix, classification_report

# Example: Suppose y_true and y_pred_classes are your true and predicted labels
# For demonstration, we'll simulate them:
num_classes = 5
np.random.seed(42)
y_true = np.random.randint(0, num_classes, 100)
y_pred_classes = np.random.randint(0, num_classes, 100)

# Generate the confusion matrix
cm = confusion_matrix(y_true, y_pred_classes)

# Convert to DataFrame for easier plotting
cm_df = pd.DataFrame(cm, 
                     index=[f"Class_{i}" for i in range(num_classes)], 
                     columns=[f"Class_{i}" for i in range(num_classes)])

# Print classification report (optional)
print("Classification Report:\n", classification_report(y_true, y_pred_classes))

# Create a confusion matrix heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(cm_df, annot=True, cmap='Blues', fmt='d')
plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")

# Save the confusion matrix to PDF
plt.savefig("confusion_matrix.pdf", bbox_inches='tight')

# Also display the plot inline (for Jupyter)
plt.show()

print("Confusion matrix PDF saved as 'confusion_matrix.pdf'.")
```
